```python
import math
import yfinance as yf
import pandas as pd
from sklearn.linear_model import LinearRegression

stock = yf.Ticker("NOK")
stock_history = stock.history(period = "5y")
stock_history.reset_index(inplace=True)

```

```python
class feature_engineer:
    def __init__(self , stock_data):
        self.stock_data = stock_data
    def calculate_five_day_moving_average(self):
        
        five_day_MA = self.stock_data["Close"].rolling(5).mean()
        return five_day_MA
    
    def calculate_20_day_moving_average(self):

        twenty_day_MA = self.stock_data["Close"].rolling(20).mean()
        return twenty_day_MA

    def calculate_50_day_moving_average(self):

        fifty_day_MA = self.stock_data["Close"].rolling(50).mean()
        return fifty_day_MA

    def calculate_daily_return(self):

        daily_return = self.stock_data["Close"].pct_change()
        return daily_return

    def twenty_day_volatility(self):

        twenty_day_volatility = self.stock_data["Close"].pct_change().rolling(20).std()
        return twenty_day_volatility

    def high_low_range(self):

        high_low = self.stock_data["High"]-self.stock_data["Low"]
        return high_low

    def create_features(self):
        features = pd.DataFrame()

        features["close"] = self.stock_data["Close"]
        features["five_day_MA"] = self.calculate_five_day_moving_average()
        features["twenty_day_MA"] = self.calculate_20_day_moving_average()
        features["fifty_day_MA"] = self.calculate_50_day_moving_average()
        features["daily_return"] = self.calculate_daily_return()
        features["volume"] = self.stock_data["Volume"]
        features["twenty_day_volatility"] = self.twenty_day_volatility()
        features["high_low_range"] = self.high_low_range()

        

        features = features.dropna()

        return features

    def create_target(self):
        target = self.stock_data["Close"].shift(-1)
        return target



        
```

```python
class model:
    def __init__(self , X , Y):
        self.X = X
        self.Y = Y
    def split_data(self):

        split_percentage = 0.8
        split_idx = math.ceil(len(self.X) * split_percentage)

        self.X_train_data = self.X.iloc[:split_idx]
        self.X_test_data = self.X.iloc[split_idx:]
        self.Y_train_data = self.Y.iloc[:split_idx]
        self.Y_test_data = self.Y.iloc[split_idx:]

    def train(self):

        self.linear_regression = LinearRegression()
        self.linear_regression.fit(self.X_train_data , self.Y_train_data)

    def prediction(self):

        self.Y_hat = self.linear_regression.predict(self.X_test_data)
        return self.Y_hat

    def mean_absolute_error(self):
        sum = 0
        count = 0
        for i in range(len(self.Y_hat)):
            diff = abs(self.Y_test_data.iloc[i] - self.Y_hat[i])
            sum = sum + diff
            count = count + 1

        error = sum/count
        return error

    



    
        
    
```
